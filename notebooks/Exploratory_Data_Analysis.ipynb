# Correlation and Multivariate Analysis

Correlation analysis is performed to understand the relationships between different numerical variables in the dataset.

The objective of this section is to identify variables that move together, detect potential multicollinearity, and determine which financial indicators are likely to contribute most towards predicting employee financial health.

These insights will guide feature selection during machine learning model development.

## 1 Correlation Heatmap

A correlation heatmap provides a visual representation of linear relationships between numerical variables.

Correlation coefficients range from **-1 to +1**, where:

- **+1** indicates a perfect positive relationship.
- **-1** indicates a perfect negative relationship.
- **0** indicates no linear relationship.

The heatmap helps identify highly related variables and possible redundancy among features.

```python
numeric_df = df.select_dtypes(include="number")

plt.figure(figsize=(18,12))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.show()
```

### Observations

- The correlation heatmap reveals several strong relationships among financial variables, indicating meaningful dependencies within employee financial behaviour.
- Salary-related variables such as Monthly Salary, Net Salary, Income Tax, and PF Contribution exhibit very high positive correlations because they are derived from the same payroll structure.
- Savings-related variables are positively correlated with one another, while expense-related variables tend to exhibit negative relationships with savings indicators.
- Expense Ratio shows a strong inverse relationship with Savings Ratio, highlighting the trade-off between spending and saving behaviour.
- The heatmap indicates the presence of multicollinearity among certain payroll variables, suggesting that feature selection or feature importance analysis should be considered before model training.
- Overall, behavioural financial metrics provide stronger and more interpretable relationships than demographic variables such as Age and Years of Experience.

 ## 2 Strongest Positive Correlations

This analysisidentifies the strongest positive relationships among numerical variables.

Understanding these relationships helps explain employee financial behaviour and may indicate variables carrying similar information.

```python
corr_matrix = numeric_df.corr()

corr_pairs = (
    corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    .stack()
    .sort_values(ascending=False)
)

corr_pairs.head(15)
```

### Observations

- Payroll-related variables exhibit extremely high positive correlations because statutory deductions and net salary are directly derived from monthly salary.
- Monthly Salary and Net Salary are almost perfectly correlated (≈0.999), indicating that higher gross income generally results in proportionally higher disposable income.
- Age and Years of Experience show a strong positive relationship (≈0.92), reflecting realistic employee career progression.
- Investments demonstrate a very strong relationship with Investment Ratio (≈0.90), confirming that employees investing larger amounts allocate a higher proportion of their income towards investments.
- The presence of highly correlated payroll variables suggests redundancy, making feature selection beneficial for improving model interpretability.

## 3 Strongest Negative Correlations

Negative correlations indicate variables that move in opposite directions.

These relationships often reveal financial trade-offs such as higher spending reducing savings or increasing debt burden.

```python
corr_pairs.tail(15)
```

### Observations

- Expense Ratio and Savings Ratio exhibit the strongest inverse relationship (≈−0.60), indicating that employees spending a larger proportion of their income generally save substantially less.
- Higher expense levels are consistently associated with lower savings and reduced investment behaviour.
- Savings Amount decreases as Expense Ratio increases, reinforcing that excessive spending limits wealth accumulation.
- Employees with higher EMI burdens tend to allocate a larger share of income toward debt repayment, reducing financial flexibility.
- These relationships demonstrate the natural financial trade-off between spending, saving, investing, and debt management.

# Feature Importance Intuition

Before training any machine learning model, it is useful to identify variables that naturally distinguish employees with different financial health levels.

Rather than relying on algorithmic feature importance, this section explores how key financial indicators vary across the Low, Medium, and High financial health categories.

The objective is to develop an intuitive understanding of which variables are likely to contribute most towards predicting employee financial health.

## 1 Average Financial Behaviour Across Financial Health Categories

The average values of key financial indicators are compared across the three financial health categories.

Large differences between the groups indicate that the corresponding variables may be informative predictors during machine learning.

```python
important_features = [
    "Monthly_Salary",
    "Net_Salary",
    "Expense_Ratio",
    "Savings_Ratio",
    "Investment_Ratio",
    "EMI_Ratio",
    "Savings_Amount",
    "Investments"
]

feature_summary = (
    df.groupby("Class")[important_features]
      .mean()
      .round(3)
)

feature_summary
```

### Observations

- Employees in the **High** financial health category have the highest average monthly salary (₹96,231) and net salary (₹81,011), while employees in the **Low** category earn slightly lower salaries (₹90,003 and ₹76,160 respectively). However, the salary differences across the three classes are relatively modest.

- A much stronger distinction is observed in behavioural financial indicators. Employees in the High financial health category spend only **35.4%** of their disposable income on expenses, compared to **51.5%** for Medium and **67.3%** for Low financial health employees.

- Savings behaviour improves substantially with financial health. The average Savings Ratio increases from **4.8%** in the Low category to **16.6%** in the Medium category and reaches **30.2%** for employees with High financial health.

- These findings suggest that employee financial wellness is influenced more by spending and saving behaviour than by salary alone, supporting earlier correlation analysis where Expense Ratio and Savings Ratio exhibited the strongest relationships with Financial Health.

## 2 Heatmap of Average Financial Indicators

A heatmap provides a compact visualization of how important financial variables differ across the financial health categories.

Brighter or darker colours highlight variables that show larger differences between classes.

```python
plt.figure(figsize=(12,6))

sns.heatmap(
    feature_summary,
    annot=True,
    cmap="YlGnBu",
    fmt=".2f"
)

plt.title("Average Financial Metrics by Financial Health")

plt.show()
```

### Observations

- The heatmap clearly demonstrates progressive improvement in employee financial behaviour from the Low to High financial health categories.

- Savings Amount, Savings Ratio, Investments, and Investment Ratio consistently increase across the three classes, indicating that financially healthier employees allocate a greater portion of their income towards long-term wealth creation.

- Expense Ratio and EMI Ratio display the opposite trend, with employees in the Low financial health category allocating a significantly larger proportion of their income towards day-to-day expenditure and debt repayment.

- Compared to behavioural variables, Monthly Salary and Net Salary exhibit relatively small variation across the three classes, reinforcing that financial discipline has a stronger influence on financial health than income itself.

Behavioural financial indicators are expected to provide stronger predictive power than demographic or salary-related variables because they directly represent employee financial habits.

The following variables are compared:

- Expense Ratio

## 4 Variables Showing the Largest Differences

The variability of each feature across the three financial health categories is calculated.

Variables exhibiting larger differences are expected to possess greater discriminatory power during predictive modelling.

```python
ranking = (
    feature_summary.max() -
    feature_summary.min()
).sort_values(ascending=False)

ranking
```

### Observations

- Savings Amount exhibits the largest variation across the financial health categories, indicating that accumulated savings provide the strongest practical distinction between financially healthy and financially stressed employees.

- Investments also demonstrate substantial variation, highlighting the importance of long-term investment behaviour in achieving better financial health.

- Although Monthly Salary and Net Salary differ across the three classes, the magnitude of these differences is considerably smaller than the behavioural differences identified in earlier analyses.

- Ratio-based variables such as Expense Ratio, Savings Ratio, and EMI Ratio provide consistent separation across the financial health categories, making them valuable predictors despite their smaller numerical ranges.

- Overall, the ranking confirms that behavioural financial metrics contribute more towards distinguishing financial health than payroll variables alone.

## 2 Outlier Detection using the Interquartile Range (IQR)

The Interquartile Range (IQR) method is used to identify observations that fall significantly above or below the normal range.

This approach provides a quantitative estimate of the number of potential outliers in each variable.

```python
def detect_outliers(series):

    q1 = series.quantile(0.25)

    q3 = series.quantile(0.75)

    iqr = q3 - q1

    lower = q1 - 1.5 * iqr

    upper = q3 + 1.5 * iqr

    return ((series < lower) | (series > upper)).sum()


for col in outlier_cols:

    print(f"{col:<25} : {detect_outliers(df[col])}")
```

### Observations

- The IQR method successfully identifies statistical outliers across several financial variables while preserving the overall structure of the dataset.

- The majority of detected outliers correspond to high-income employees, employees with large savings balances, or individuals making substantial investments.

- These observations reflect genuine financial diversity within the synthetic workforce and are expected based on the salary generation rules implemented during data creation.

- Consequently, the detected outliers should be interpreted as valid employee profiles rather than anomalies requiring removal.

```python
outlier_summary = []

for col in outlier_cols:

    count = detect_outliers(df[col])

    percentage = round(count / len(df) * 100,2)

    outlier_summary.append([col,count,percentage])

outlier_df = pd.DataFrame(

    outlier_summary,

    columns=["Feature","Outliers","Percentage"]

)

outlier_df
```

### Observations

- The percentage of outliers varies across the selected financial variables, indicating differing levels of variability within employee financial behaviour.

- Variables representing absolute monetary values generally contain a higher proportion of outliers than ratio-based variables because employee income, savings, and investment capacities differ considerably across departments and seniority levels.

- Ratio-based features remain comparatively stable despite substantial variation in salary, demonstrating their effectiveness in normalising financial behaviour across employees.

- The observed outlier percentages remain within realistic limits for a heterogeneous corporate workforce and do not indicate abnormal data generation.
